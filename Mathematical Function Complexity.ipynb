## **How to Rank the Functions**  

When ranking, use these **common Big O categories** as a reference:  

- **Constant Time:** Functions that do not depend on $ n $ (e.g., $ O(1) $).
- **Logarithmic Time:** Functions that grow logarithmically (e.g., $ O(\log n) $).
- **Linear Time:** Functions that grow linearly (e.g., $ O(n) $).
- **Linearithmic Time:** Functions that grow as $ n \log n $.
- **Polynomial Time:** Functions that grow polynomially (e.g., $ O(n^k) $).
- **Exponential Time:** Functions that grow exponentially (e.g., $ O(2^n) $).
- **Factorial Time:** Functions that grow factorially (e.g., $ O(n!) $).

## **My Answer: Ranking (least to most complex)**

| Rank | Function | Big O | Explanation |
|---|---|---|---|
| 1 | $f_1(n) = 2$ | $O(1)$ | Constant: the value never depends on $n$. |
| 2 | $f_2(n) = \log(n)$ | $O(\log n)$ | Logarithmic: grows extremely slowly, e.g. halving the search space each step (binary search). |
| 3 | $f_{12}(n) = \sqrt{n}$ | $O(n^{1/2})$ | Grows slower than linear but faster than any logarithm (sub-linear polynomial). |
| 4 | $f_3(n) = n$ | $O(n)$ | Linear: doubling $n$ doubles the work. |
| 4 | $f_6(n) = 3n + 5$ | $O(n)$ | Linear: constants (3 and 5) are dropped, so it ties with $f_3$. |
| 6 | $f_4(n) = n \log(n)$ | $O(n \log n)$ | Linearithmic: typical for efficient sorting (merge sort). |
| 7 | $f_5(n) = n^2$ | $O(n^2)$ | Quadratic: nested loops over the input. |
| 7 | $f_7(n) = 4n^2 + 2n + 1$ | $O(n^2)$ | Only the dominant term $n^2$ counts, so it ties with $f_5$. |
| 9 | $f_8(n) = n^3$ | $O(n^3)$ | Cubic: three nested loops (e.g. naive matrix multiplication). |
| 10 | $f_{11}(n) = 7n^4 + n^2$ | $O(n^4)$ | Polynomial of degree 4: dominated by $n^4$, the constant 7 is dropped. |
| 11 | $f_9(n) = 5^n$ | $O(5^n)$ | Exponential: beats every polynomial once $n$ is large enough. |
| 12 | $f_{10}(n) = n!$ | $O(n!)$ | Factorial: grows even faster than exponentials (e.g. brute-force route search). |

**Ranking in one line:** $1 < \log n < \sqrt{n} < n \approx 3n+5 < n\log n < n^2 \approx 4n^2+2n+1 < n^3 < 7n^4+n^2 < 5^n < n!$

**Rules used:** drop constants, keep only the fastest growing term, and remember the hierarchy constant < logarithmic < polynomial < exponential < factorial.

```python
# Check the ranking numerically: evaluate every function for a few values of n
import math

funcs = {
    "f1 = 2": lambda n: 2,
    "f2 = log n": lambda n: math.log(n),
    "f12 = sqrt n": lambda n: math.sqrt(n),
    "f3 = n": lambda n: n,
    "f6 = 3n + 5": lambda n: 3 * n + 5,
    "f4 = n log n": lambda n: n * math.log(n),
    "f5 = n^2": lambda n: n ** 2,
    "f7 = 4n^2 + 2n + 1": lambda n: 4 * n ** 2 + 2 * n + 1,
    "f8 = n^3": lambda n: n ** 3,
    "f11 = 7n^4 + n^2": lambda n: 7 * n ** 4 + n ** 2,
    "f9 = 5^n": lambda n: 5 ** n,
    "f10 = n!": lambda n: math.factorial(n),
}

n = 30
values = {name: f(n) for name, f in funcs.items()}
for name, v in sorted(values.items(), key=lambda kv: kv[1]):
    print(f"{name:<20} {v:.3e}")
```
